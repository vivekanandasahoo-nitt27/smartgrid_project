```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = final_dataset.copy()

# Remove class column for outlier analysis
features = df.drop(columns=['class'])

# Store outlier counts
outlier_counts = {}

# Plot boxplots and count outliers
for col in features.columns:
    
    # Q1 and Q3
    Q1 = features[col].quantile(0.25)
    Q3 = features[col].quantile(0.75)
    
    # IQR
    IQR = Q3 - Q1
    
    # Outlier limits
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    
    # Count outliers
    outliers = features[(features[col] < lower) | (features[col] > upper)]
    
    outlier_counts[col] = len(outliers)
    
    # Boxplot
    plt.figure(figsize=(12,4))
    plt.boxplot(features[col], vert=False)
    plt.title(f"{col} Boxplot")
    plt.grid(True)
    plt.show()

# Convert to dataframe
outlier_df = pd.DataFrame({
    'Column': outlier_counts.keys(),
    'Outlier_Count': outlier_counts.values()
})

# Sort by highest outliers
outlier_df = outlier_df.sort_values(by='Outlier_Count', ascending=False)

# Display
print(outlier_df)

# Total outliers
print("\nTotal Outliers Across All Columns:",
      outlier_df['Outlier_Count'].sum())
```

```python
import pandas as pd
import numpy as np
import joblib

# ==========================
# LOAD MODEL + FEATURES
# ==========================

model = joblib.load("smartgrid_rf_model3.pkl")
feature_cols = joblib.load("feature_columns3.pkl")

# ==========================
# LOAD TEST DATA
# ==========================

df = pd.read_csv("fault_test10.csv")

# Keep only model features
X = df[feature_cols].copy()

# ==========================
# ORIGINAL PREDICTION
# ==========================

print("\n========================")
print("ORIGINAL PREDICTION")
print("========================")

orig_pred = model.predict(X)
orig_prob = model.predict_proba(X)

for i in range(30,35):

    print(f"\nRow {i}")

    print("Prediction :", orig_pred[i])

    print("Confidence :", np.max(orig_prob[i]))

    print("Probabilities :", orig_prob[i])

# ==========================
# GET MOST IMPORTANT FEATURES
# ==========================

importances = model.feature_importances_

importance_df = pd.DataFrame({
    'Feature': feature_cols,
    'Importance': importances
})

importance_df = importance_df.sort_values(
    by='Importance',
    ascending=False
)

print("\n========================")
print("TOP 10 IMPORTANT FEATURES")
print("========================")

print(importance_df.head(10))

top10_features = importance_df.head(10)['Feature'].tolist()

# ==========================
# CREATE ATTACKED COPY
# ==========================

X_attack = X.copy()

# Change ONLY row 50
target_row = 50

print("\n========================")
print("MODIFYING ROW 50")
print("========================")

for feat in top10_features:

    original_value = X_attack.loc[target_row, feat]

    
    attack_multiplier = np.random.uniform(1.9, 2.8)

    attacked_value = original_value * attack_multiplier

    X_attack.loc[target_row, feat] = attacked_value

    print(f"{feat}")
    print(f"Original : {original_value}")
    print(f"Attacked : {attacked_value}")
    print()

# ==========================
# PREDICT AFTER ATTACK
# ==========================

print("\n========================")
print("AFTER ATTACK")
print("========================")

attack_pred = model.predict(X_attack)
attack_prob = model.predict_proba(X_attack)

print("\nROW 50 RESULT")

print("\nOriginal Prediction:")
print(orig_pred[target_row])

print("Original Confidence:")
print(np.max(orig_prob[target_row]))

print("Original Probabilities:")
print(orig_prob[target_row])

print("\nAfter Attack Prediction:")
print(attack_pred[target_row])

print("After Attack Confidence:")
print(np.max(attack_prob[target_row]))

print("After Attack Probabilities:")
print(attack_prob[target_row])



orig_conf = np.max(orig_prob[target_row])
new_conf = np.max(attack_prob[target_row])

print("\n========================")
print("CONFIDENCE CHANGE")
print("========================")

print("Original Confidence :", orig_conf)
print("New Confidence      :", new_conf)
print("Confidence Drop     :", orig_conf - new_conf)
```

```python
import pandas as pd
import numpy as np
import joblib
import random
import matplotlib.pyplot as plt



model = joblib.load("smartgrid_rf_model3.pkl")

feature_cols = joblib.load("feature_columns3.pkl")

df = pd.read_csv("fault_test11.csv")

X = df[feature_cols].copy()



current_features = [
    'threePhaseCMUA1', 'threePhaseCMUB1', 'threePhaseCMUC1',
    'threePhaseCMUA2', 'threePhaseCMUB2', 'threePhaseCMUC2',
    'threePhaseCMUA3', 'threePhaseCMUB3', 'threePhaseCMUC3',
    'threePhaseCMUA4', 'threePhaseCMUB4', 'threePhaseCMUC4'
]

voltage_features = [
    'threePhaseVMUA1', 'threePhaseVMUB1', 'threePhaseVMUC1',
    'threePhaseVMUA2', 'threePhaseVMUB2', 'threePhaseVMUC2',
    'threePhaseVMUA3', 'threePhaseVMUB3', 'threePhaseVMUC3',
    'threePhaseVMUA4', 'threePhaseVMUB4', 'threePhaseVMUC4'
]

freq_features = [
    'frequencyMU1',
    'frequencyMU2',
    'frequencyMU3',
    'frequencyMU4'
]



X_attack = X.copy()

n_rows = len(X_attack)

print("\n================================")
print("GENERATING REALISTIC ATTACKS")
print("================================")



for row in range(n_rows - 20):

    attack_type = random.random()

    

    if attack_type < 0.35:

        # Voltage drop attack first
        num_v = random.randint(1, 4)

        v_feats = random.sample(voltage_features, num_v)

        for feat in v_feats:

            drop = random.uniform(0.55, 0.92)

            X_attack.loc[row, feat] *= drop

        # Small overlap with current attack
        if random.random() < 0.40:

            c_feats = random.sample(current_features,
                                    random.randint(1, 2))

            for feat in c_feats:

                spike = random.uniform(1.2, 2.0)

                X_attack.loc[row, feat] *= spike



    elif attack_type < 0.65:

        num_c = random.randint(1, 4)

        c_feats = random.sample(current_features, num_c)

        for feat in c_feats:

            spike = random.uniform(1.8, 5.0)

            X_attack.loc[row, feat] *= spike

        # Small voltage overlap
        if random.random() < 0.30:

            v_feats = random.sample(voltage_features,
                                    random.randint(1, 2))

            for feat in v_feats:

                X_attack.loc[row, feat] *= random.uniform(0.75, 0.95)

    

    elif attack_type < 0.85:

        f_feats = random.sample(freq_features,
                                random.randint(1, 3))

        for feat in f_feats:

            noise = (
                0.4 * np.sin(row / random.uniform(2, 8))
                +
                np.random.normal(0, 0.15)
            )

            X_attack.loc[row, feat] += noise

        # Sometimes overlap with voltage
        if random.random() < 0.35:

            v_feats = random.sample(voltage_features,
                                    random.randint(1, 2))

            for feat in v_feats:

                X_attack.loc[row, feat] *= random.uniform(0.7, 0.9)

    

    else:

        # Voltage collapse
        for feat in random.sample(voltage_features, 4):

            X_attack.loc[row, feat] *= random.uniform(0.3, 0.7)

        # Current surge
        for feat in random.sample(current_features, 4):

            X_attack.loc[row, feat] *= random.uniform(2.5, 5.5)

        # Frequency distortion
        for feat in random.sample(freq_features, 2):

            X_attack.loc[row, feat] += (
                np.sin(row)
                +
                np.random.normal(0, 0.3)
            )



print("\n================================")
print("ADDING MASSIVE FINAL ATTACK")
print("================================")

for row in range(n_rows - 20, n_rows):

    # Voltage collapse
    for feat in voltage_features:

        X_attack.loc[row, feat] *= random.uniform(0.4, 0.9)

    # Current explosion
    for feat in current_features:

        X_attack.loc[row, feat] *= random.uniform(1.0, 5.0)

    # Frequency chaos
    for feat in freq_features:

        X_attack.loc[row, feat] += (
            3 * np.sin(row)
            +
            np.random.normal(0, 1)
        )



print("\n================================")
print("RUNNING MODEL")
print("================================")

pred = model.predict(X_attack)

prob = model.predict_proba(X_attack)

confidence = np.max(prob, axis=1)



results_df = X_attack.copy()

results_df['Prediction'] = pred

results_df['Confidence'] = confidence

# Save probabilities
for i in range(prob.shape[1]):

    results_df[f'Class_{i}_Prob'] = prob[:, i]

results_df.to_csv(
    "realistic_cyberattack_dataset.csv",
    index=False
)

print("\nDataset saved as:")
print("realistic_cyberattack_dataset.csv")



threshold = 0.70

low_conf_idx = np.where(confidence < threshold)[0]

print("\n================================")
print("LOW CONFIDENCE EVENTS")
print("================================")

print("Total Low Confidence Rows :", len(low_conf_idx))



plt.figure(figsize=(15,5))

plt.plot(confidence)

plt.axhline(
    y=threshold,
    linestyle='--'
)

plt.title("Prediction Confidence")

plt.xlabel("Row")

plt.ylabel("Confidence")

plt.grid(True)

plt.show()



for i in range(prob.shape[1]):

    plt.figure(figsize=(15,5))

    plt.plot(prob[:, i])

    plt.title(f"Class {i} Probability")

    plt.xlabel("Row")

    plt.ylabel("Probability")

    plt.grid(True)

    plt.show()


print("\n================================")
print("SAMPLE RESULTS")
print("================================")

for i in range(10):

    print(f"\nRow {i}")

    print("Prediction :", pred[i])

    print("Confidence :", confidence[i])

    print("Probabilities :", prob[i])
    
    
    

from matplotlib.backends.backend_pdf import PdfPages
import matplotlib.pyplot as plt

pdf_name = "model_prediction_graphs.pdf"

with PdfPages(pdf_name) as pdf:

    # ==========================================
    # 1. Prediction Confidence Graph
    # ==========================================

    plt.figure(figsize=(15, 5))

    plt.plot(confidence)

    plt.axhline(
        y=threshold,
        linestyle='--',
        label=f"Threshold = {threshold}"
    )

    plt.title("Prediction Confidence")
    plt.xlabel("Row")
    plt.ylabel("Confidence")

    plt.legend()
    plt.grid(True)

    pdf.savefig(bbox_inches='tight')
    plt.close()


    # ==========================================
    # 2. Probability Graph for Each Class
    # ==========================================

    for i in range(prob.shape[1]):

        plt.figure(figsize=(15, 5))

        plt.plot(prob[:, i])

        plt.title(f"Class {i} Probability")

        plt.xlabel("Row")
        plt.ylabel("Probability")

        plt.ylim(0, 1)

        plt.grid(True)

        pdf.savefig(bbox_inches='tight')
        plt.close()


print("================================")
print("PDF CREATED SUCCESSFULLY")
print("================================")
print(f"File: {pdf_name}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Number of samples
N = 300
t = np.arange(N)

# ----------------------------
# Normal Signals
# ----------------------------
voltage = 1.0 + 0.01*np.sin(2*np.pi*t/40)

current = 0.85 + 0.03*np.sin(2*np.pi*t/35)

frequency = 50 + 0.02*np.sin(2*np.pi*t/60)

# ----------------------------
# Attack starts here
# ----------------------------
attack_start = 180

voltage_attack = voltage.copy()
current_attack = current.copy()
frequency_attack = frequency.copy()

# Voltage collapse + oscillation + noise
voltage_attack[attack_start:] = (
    voltage_attack[attack_start:]
    - 0.35
    + 0.08*np.sin(2*np.pi*np.arange(N-attack_start)/3)
    + np.random.normal(0,0.02,N-attack_start)
)

# Current surge + oscillation + noise
current_attack[attack_start:] = (
    current_attack[attack_start:]
    + 2.5
    + 0.45*np.sin(2*np.pi*np.arange(N-attack_start)/4)
    + np.random.normal(0,0.08,N-attack_start)
)

# Frequency distortion
frequency_attack[attack_start:] = (
    50
    + 0.8*np.sin(2*np.pi*np.arange(N-attack_start)/2)
    + 0.5*np.sin(2*np.pi*np.arange(N-attack_start)/7)
    + np.random.normal(0,0.12,N-attack_start)
)

# ----------------------------
# Severe attack in last 20 samples
# ----------------------------
voltage_attack[-20:] -= 0.25*np.random.rand(20)

current_attack[-20:] += 2*np.random.rand(20)

frequency_attack[-20:] += 2*np.sin(np.arange(20)*3)

# ----------------------------
# Plot
# ----------------------------

fig, ax = plt.subplots(3,1,figsize=(10,8),sharex=True)

# Voltage
ax[0].plot(t, voltage, '--', linewidth=1.5, label='Normal')
ax[0].plot(t, voltage_attack, linewidth=2, label='Under Voltage Attack')
ax[0].axvline(attack_start, color='red', linestyle='--')
ax[0].set_title("Voltage Collapse with Oscillatory Cyberattack")
ax[0].set_ylabel("Voltage (p.u.)")
ax[0].grid(True)
ax[0].legend()

# Current
ax[1].plot(t, current, '--', linewidth=1.5, label='Normal')
ax[1].plot(t, current_attack, linewidth=2, label='Overcurrent Attack')
ax[1].axvline(attack_start, color='red', linestyle='--')
ax[1].set_title("Current Surge Attack")
ax[1].set_ylabel("Current (A)")
ax[1].grid(True)
ax[1].legend()

# Frequency
ax[2].plot(t, frequency, '--', linewidth=1.5, label='Normal')
ax[2].plot(t, frequency_attack, linewidth=2, label='Frequency Manipulation')
ax[2].axvline(attack_start, color='red', linestyle='--')
ax[2].set_title("Frequency Distortion Attack")
ax[2].set_ylabel("Frequency (Hz)")
ax[2].set_xlabel("Time Step")
ax[2].grid(True)
ax[2].legend()

plt.tight_layout()
plt.show()
```
