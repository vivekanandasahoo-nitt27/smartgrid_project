```python
import pandas as pd
import numpy as np
import random



attack_df = pd.read_csv('attack_V_I_freq_angle.csv')
fault_df = pd.read_csv(r'D:\PROJECTS\smartgrid_project\SYSTEM\combined_fault_dataset.csv')
healthy_df = pd.read_csv('healthy_dataset_complete.csv')


cols = [
    'IED4_iA', 'IED4_iB', 'IED4_iC',
    'MU1CurrentAngleA', 'MU1CurrentAngleB', 'MU1CurrentAngleC',
    'MU1VoltageAngleA', 'MU1VoltageAngleB', 'MU1VoltageAngleC',

    'MU2CurrentAngleA', 'MU2CurrentAngleB', 'MU2CurrentAngleC',
    'MU2VoltageAngleA', 'MU2VoltageAngleB', 'MU2VoltageAngleC',

    'MU3CurrentAngleA', 'MU3CurrentAngleB', 'MU3CurrentAngleC',
    'MU3VoltageAngleA', 'MU3VoltageAngleB', 'MU3VoltageAngleC',

    'MU4VoltageAngleA', 'MU4VoltageAngleB', 'MU4VoltageAngleC',

    'consistency_1',
    'consistency_2',
    'consistency_3',

    'frequencyMU1',
    'frequencyMU2',
    'frequencyMU3',
    'frequencyMU4',

    'sqNum',

    'threePhaseCMUA2',
    'threePhaseCMUB2',
    'threePhaseCMUC2',

    'threePhaseCMUA3',
    'threePhaseCMUB3',
    'threePhaseCMUC3',

    'threePhaseCMUA4',
    'threePhaseCMUB4',
    'threePhaseCMUC4',

    'threePhaseCSum_1',
    'threePhaseCSum_2',
    'threePhaseCSum_3',

    'threePhaseCurrentMUA1',
    'threePhaseCurrentMUB1',
    'threePhaseCurrentMUC1',

    'threePhaseVMUA1',
    'threePhaseVMUB1',
    'threePhaseVMUC1',

    'threePhaseVMUA2',
    'threePhaseVMUB2',
    'threePhaseVMUC2',

    'threePhaseVMUA3',
    'threePhaseVMUB3',
    'threePhaseVMUC3',

    'threePhaseVMUA4',
    'threePhaseVMUB4',
    'threePhaseVMUC4',

    'threePhaseVSum_1',
    'threePhaseVSum_2',
    'threePhaseVSum_3',

    'class'
]



attack_df = attack_df[cols].copy()
fault_df = fault_df[cols].copy()
healthy_df = healthy_df[cols].copy()



attack_df.reset_index(drop=True, inplace=True)
fault_df.reset_index(drop=True, inplace=True)
healthy_df.reset_index(drop=True, inplace=True)



final_chunks = []

attack_idx = 0
fault_idx = 0
healthy_idx = 0

attack_len = len(attack_df)
fault_len = len(fault_df)
healthy_len = len(healthy_df)

event_counter = 0



while (
    attack_idx < attack_len or
    fault_idx < fault_len or
    healthy_idx < healthy_len
):

    

    if healthy_idx < healthy_len:

        healthy_block_size = random.randint(80, 200)

        healthy_block = healthy_df.iloc[
            healthy_idx:
            healthy_idx + healthy_block_size
        ].copy()

        healthy_block['scenario_id'] = f'healthy_{event_counter}'

        final_chunks.append(healthy_block)

        healthy_idx += healthy_block_size

    

    event_type = random.choice(['fault', 'attack'])

    

    if (
        event_type == 'fault' and
        fault_idx < fault_len
    ):

        fault_block_size = random.randint(20, 60)

        fault_block = fault_df.iloc[
            fault_idx:
            fault_idx + fault_block_size
        ].copy()

        fault_block['scenario_id'] = f'fault_{event_counter}'

        final_chunks.append(fault_block)

        fault_idx += fault_block_size

    

    elif (
        event_type == 'attack' and
        attack_idx < attack_len
    ):

        attack_block_size = random.randint(40, 120)

        attack_block = attack_df.iloc[
            attack_idx:
            attack_idx + attack_block_size
        ].copy()

        attack_block['scenario_id'] = f'attack_{event_counter}'

        final_chunks.append(attack_block)

        attack_idx += attack_block_size

    event_counter += 1



Train_14_bus_dataset = pd.concat(
    final_chunks,
    ignore_index=True
)



Train_14_bus_dataset['time'] = (
    np.arange(len(Train_14_bus_dataset)) * 0.2
)



Train_14_bus_dataset.reset_index(
    drop=True,
    inplace=True
)



Train_14_bus_dataset.to_csv(
    'Train_14_bus_dataset.csv',
    index=False
)



print(Train_14_bus_dataset.shape)

print("\nClass Distribution:\n")
print(Train_14_bus_dataset['class'].value_counts())

print("\nScenario Examples:\n")
print(
    Train_14_bus_dataset['scenario_id']
    .head(20)
)
```
